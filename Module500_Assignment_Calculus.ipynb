-  **Step 2**:  
Write a function that implements gradient descent from scratch (i.e. you may not call any library functions) to generate the linear regression equation coefficients by minimizing the cost function.
Here we use the Mean Square Error (MSE) cost function where n is the number of examples in the dataset, y is the dependent variable in the dataset, $c_0$ is the y-intercept for the linear equation, $c_1$ is the slope of the linear equation, and $x_i$ is the independent variable ($y = c_1x_i + c_0$). Note that in the right-hand side of the equation below, $\widehat y$, the estimate for the dependent variable, has been replaced with the specific iteration of the regression equation, $ c_0 + c_1x_i$.

$$MSE = \frac{1}{n}\sum(y - \widehat y)^2 = \frac{1}{n}\sum(y - (c_0+c_1x_i))^2$$  

           Function API:  def gradientDescent(coeffs, X, y, μ, iterations):
           Input: coefficients: array of coefficients (i.e. c0 and c1) initialized to random values
                  X: normalized training dataset independent variable/s (see hint below)
                  y: normalized training dataset dependent variable (see hint below)
                  μ:  learning rate  
                  iterations: number of iterations to run
           Output: Returns c0_arr, c1_arr, MSE_arr (in this order), where c0_arr, c1_arr, MSE_arr are numpy arrays
                   of the c0, c1 coefficients and sum-of-squared error respectively for each of the iterations performed in your gradient descent loop.

The following are the partial derivatives of the cost function with respect to each of the coefficients. These are the "gradients".

$$\frac{\partial(MSE)}{\partial c_0} = \frac{1}{n}\cdot 2(y-(c_0+c_1x_i))\cdot -1 = -\frac{2}{n}(y-\widehat y) $$

 $$\frac{\partial(MSE)}{\partial c_1} = \frac{1}{n}\cdot 2(y-(c_0+c_1x_i))\cdot -x_i = -\frac{2}{n}(y-\widehat y)x_i $$
> Hint:  
 - The following calculations are done over each sample in the training set (i.e. use a for loop).
 - Calc the model prediction:  
    $\widehat y = c_0 + c_1x_i$
 - Calc the MSE:   
    $MSE = \frac{1}{n}\sum(y - \widehat y)^2$
 - Calc the gradient of c0:  
    $\frac{\partial(MSE)}{\partial c_0}=-\frac{2}{n}\sum(y-\widehat y)$
 - Calc the gradient of c1:  
    $\frac{\partial(MSE)}{\partial c_1}=-\frac{2}{n}\sum(y-\widehat y)x_i$
 - Update the coefficients:  
    - c0 = c0 - μ(gradient of c0)  
    - Append calculated c0 to c0 array
    - c1 = c1 - μ(gradient of c1)  
    - Append calculated c1 to c1 array

```
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
import random
from sklearn.metrics import r2_score


#Input Parameters
Age = np.array([43,21,25,42,57,59,35,15,55,50,65,10,45,35])
Glucose = np.array([99,65,79,75,87,81,80,80,90,70,95,67,90,82])


#Step 1:  Create a function which MANUALLY (using the equations) calculates the
#         coefficients for a simple linear regression model, then plot the calculated
#         regression line on top of a scatterplot of the data.
#         Your function must input numpy arrays for the x and y variables and return c0 and c1.
#         Your function MUST use equations shown in step 1 of the text cell above.
#         Note - you may not use a machine learning library model for this step except numpy arrays and operations.
#         Note - this will be the same as the function you created in the Linear Prediction Project.
#         Note - normalize your data using the min/max method before calculating the coefficients.

#Normalize the data using the following min/max method:
#Normalized Sample = (Sample - Sample.min)/(Sample.max - Sample.min)
# YOUR CODE HERE...

X = (Age - Age.min())/(Age.max() - Age.min())
y = (Glucose - Glucose.min())/(Glucose.max() - Glucose.min())


#Write a function to calculate the Linear Regression coefficients manually.
#See the text cell notes for the equations
def simpleLRcoeffsManual(X, y):
# YOUR CODE HERE...
  coeff_c1 = np.sum((X - np.mean(X))* (y - np.mean(y)))/np.sum((X - np.mean(X))**2)
  coeff_c0 = np.mean(y) - coeff_c1 * np.mean(X)

  return coeff_c0,coeff_c1


#Call your function to generate the model coefficients
# YOUR CODE HERE...
c0_direct, c1_direct = simpleLRcoeffsManual(X, y)

#Generate the prediction outputs
ypred_direct = c0_direct + c1_direct * X
print('ypred_direct:', ypred_direct)

#Calculate the MSE manually
#Use the left side of the MSE formula as shown at the top of step 2
MSE_direct = np.sum((y - ypred_direct)**2) / len(y)


#Plot the data and prediction line for the manual calculation
# YOUR CODE HERE...
plt.plot(X, y, 'o')
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.plot(X, ypred_direct, 'b--' )
plt.show()


import doctest

'''
  >>> #Direct coefficients
  >>> print(np.round(ypred_direct[0], 3))
  0.517
  >>> print(np.round(ypred_direct[-1], 3))
  0.433
  >>> print(np.round(MSE_direct, 3))
  0.053
'''

doctest.testmod()
```

```


#Step 2:  Write a function that implements gradient descent to minimize the cost function.
#         where the cost function is sum-of-square error (MSE)
#         Inputs:  coeffs:  array of coefficients c0 and c1 initialized to random values
#                  X:  dataset normalized independent variable values
#                  y:  dataset normalized dependent variable values
#                  lr: learning rate
#                  iterations:  number of iterations
#         Outputs: Returns c0_arr, c1_arr, MSE_arr (in this order)
#         def gradientDescent(coeffs, X, y, lr, iterations):
#
def gradientDescent(coeffs, X, y, lr, iterations):
# YOUR CODE HERE...

  #Init c0 and c1 to the initial input parameter coeffs
  c0, c1 = coeffs

  #Create empty numpy arrays to hold coeffs and MSE over each iteration (but empty for now)
  c1_arr = np.array([])
  c0_arr = np.array([])
  MSE_arr = np.array([])

  #Loop over the number of iterations
  for i in range(iterations):

    #Calculate the yhat vector based on the current iteration's coefficients
    yhat = c0 + c1 * X

    #Calculate MSE of current yhat and y values. We will be minimizing this value
    #over the course of all the iterations
    MSE = (1/len(X)) * np.sum((y - yhat)**2)

    #Calculate gradients for the trainable parameters (coefficients c0, c1)
    grad_c0 = (-2/len(X)) * np.sum(y - yhat)
    grad_c1 = (-2/len(X)) * np.sum((y - yhat)*X)

    #Update the trainable parameters (coefficients c0, c1) using the learning rate and the gradients
    #We should be getting closer to the true values with each iteration
    c0 = c0 - grad_c0 * lr
    c1 = c1 - grad_c1 * lr

    #Append the coefficients and MSE to arrays (for plotting later) using np.append
    c0_arr = np.append(c0_arr, c0)
    c1_arr = np.append(c1_arr, c1)
    MSE_arr = np.append(MSE_arr, MSE)

  #return the three np arrays so we can trace our progress in gradient descent
  return c0_arr, c1_arr, MSE_arr


# Step 3: Plot the equation of your model from the manual (Step 1) model and the gradient descent (Step 2) model
#         on the same graph along with your dataset points.
#         Note, Use the normalized datasets you created in Step 1.
#         Run your algorithm for 100 epochs (100 iterations)
#         Set your learning rate (lr) to 0.01

epochs = 100                        #for test - do not change - number of repetitions of the algorithm
initial_c = np.array([0.4, 0.7])    #for test - do not change - Initial guesses for the two coefficients
lr = 0.01                           #for test - do not change - Learning Rate

# YOUR CODE HERE...

#Call the gradientDescent function on your normalized X and y data from Step 1
coef_c0, coef_c1, error = gradientDescent(initial_c, X, y, lr, epochs)

#Prints each numpy array returned from your function call
print('descent, coef_c0 over 100 iterations:', coef_c0)
print('descent, coef_c1 over 100 iterations:', coef_c1)
print('descent, mse (mean squared errors) cost function over 100 iterations:', error)

#Calculate the predictions (yPred) using trained coefficients
#   Hint: use the last coefficients in the returned arrays
#         Also don't forget to use the normed data
yPred = coef_c0[-1] + coef_c1[-1] * X

#Graphs the output on the same plot as the manual prediction (you can copy the same
#plot from above but now include the new yPred as a red dashed line)
plt.plot(X, y, 'o')
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.plot(X, ypred_direct, 'b--' )
plt.plot(X, yPred, 'r--')
plt.show()



# Step 4:  Experiment with different numbers of iterations to see how your gradient descent
#          model changes compared to that of manual model.   You should observe an improvement
#          with increased number of iterations such the two techniques converge with higher numbers
#          of iterations. (i.e. the lines should coincide).

#         You do not need to code anything here-- just try changing your number of epochs
#         above and see if your lines look closer together


#         Then, change your epochs back to 100 so that the doctest below does not fail




# Step 5: Plot on a 3D graph your coefficients (c0 and c1), on x and y axes respectively,
#         and MSE (Sum of Square Errors) on the z-axis.  Your plot should should the
#         MSE converging to a minimum as shown in the figure below.  Note your graph will
#         not look exactly as the plot shown as it will be depending on the initial
#         random values of the coefficients and the number of iterations.
#         Hint: https://matplotlib.org/2.0.2/mpl_toolkits/mplot3d/tutorial.html

# Note - The code below will generate a 3D plot
#
#         ALL CODE FROM STEP 5 WAS ALREADY PROVIDED
#         Simply try to understand what it is doing


z2D = np.empty([epochs,epochs])
for m in range(0,epochs):
  z2D[m,:] = np.array(error[m]*np.ones((epochs)))

fig = plt.figure(figsize=(20,10))   #was 10,10 in solution doc
ax = plt.axes(projection='3d')
ax.set_xlabel('c0', fontsize=12, loc='right')
ax.set_ylabel('c1', fontsize=12, loc='bottom')
ax.set_zlabel('MSE', fontsize=12 )   #fontsize was 8 in solution doc
ax1 = plt.axes(projection='3d')
ax1.scatter(coef_c0, coef_c1, error, c='b')
ax1.plot(coef_c0, coef_c1, error, color='b')

#Find the min value and index in the error array
minZ = error[0]
minZindex = 0
for n in range (0, epochs):
  if error[n] < minZ:
    minZ = error[n]
    minZindex = n
print('minZ, index:', minZ, minZindex)

#Highlight the last coefficents (should be the ones with the minimized cost)
ax1.scatter(coef_c0[epochs-1], coef_c1[epochs-1], error[epochs-1], color='r', s=100)
plt.show()


#Step 6:  Calculate the performance for both techniques (Step 2, Step 3) using the
# r-squared score metric (r2_score imported above from scikitlearn)
# YOUR CODE HERE...
r2_score_direct = r2_score(y, ypred_direct)
r2_score_gradDescent = r2_score(y, yPred)

print('Score direct calculated coeffs:', r2_score_direct)
print('Score gradient descent calculated coeffs:', r2_score_gradDescent)

import doctest

'''
  >>> #Regression coeffs trained with Gradient Descent
  >>> print(np.round(coef_c0[-1], 3))
  0.19
  >>> print(np.round(error[-1], 3))
  0.053
  >>> #R2 score - Direct coefficients
  >>> print(np.round(r2_score_direct, 3))
  0.357
  >>> #R2 score - Regression coeffs trained with Gradient Descent
  >>> print(np.round(r2_score_gradDescent, 3))
  0.352
'''

doctest.testmod()
```
